## 1. 环境检查

`sys.executable` 会告诉我们当前 notebook 内核对应的 Python 路径。你希望看到的是 `pytorch-study-notebook` 环境。

````python
import sys
import platform

print("Python 路径：", sys.executable)
print("Python 版本：", sys.version.split()[0])
print("操作系统：", platform.platform())
````

## 2. PyTorch 版本检查

如果下面这格能正常运行，说明 PyTorch 已经安装好。

````python
import torch
import torchvision
import numpy as np

print("torch 版本：", torch.__version__)
print("torchvision 版本：", torchvision.__version__)
print("numpy 版本：", np.__version__)
````

## 3. 选择计算设备

- `cuda`：NVIDIA GPU。
- `mps`：Apple Silicon Mac 的 Metal 加速。
- `cpu`：普通 CPU，兼容性最好。

入门学习阶段不用执着 GPU；能先跑通流程最重要。

````python
if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print("当前可用设备：", device)
print("CUDA 可用：", torch.cuda.is_available())
print("MPS 可用：", torch.backends.mps.is_available())
````

## 4. 最小 tensor 示例

Tensor 可以先理解成 PyTorch 里的数组/矩阵。神经网络里的数据、权重、梯度，基本都用 tensor 表示。

````python
x = torch.arange(6).reshape(2, 3)
y = torch.ones_like(x)

print("x =")
print(x)
print("y =")
print(y)
print("x + y =")
print(x + y)
print("x 的形状：", x.shape)
print("x 的数据类型：", x.dtype)
````

## 5. 安装命令怎么看

原文中的安装命令通常长这样：

```bash
pip install torch torchvision
```

在 notebook 里更推荐用：

```python
%pip install 包名
```

`%pip` 会安装到当前 notebook 内核所在的环境里，比 `!pip` 更不容易装错地方。

````python
print("如果以后需要在 notebook 中安装包，推荐写法：%pip install 包名")
print("当前环境已经配好 PyTorch，学习本仓库时一般不需要重复安装。")
````
